```python
b_pr = int(input("Купил по "))

s_pr = int(input("\nПродал по "))

print("\n")
if b_pr < s_pr:
  print("Прибыль:")
  print(f"+{s_pr - b_pr}")
elif b_pr == s_pr:
  print("Прибыль: 0")
else:
  print("Убыток:")
  print(f"-{b_pr - s_pr}")

```
